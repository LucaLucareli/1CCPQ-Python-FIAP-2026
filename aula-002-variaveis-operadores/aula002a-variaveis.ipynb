# Declaração de Dados

### Tipos Primitivos

```python
# No Notebook, podemos declarar e já ver o resultado
nome = "Luca"
print(f"Tipo do nome: {type(nome)}")

idade = 18
print(f"Tipo da idade: {type(idade)}")
```

### Tipos Compostos

```python
# O Dicionário (dict) é o "rei" das variáveis para organizar dados
perfil = {
    "nome": "Luca",
    "skills": ["Python", "Data Science", "Web"], # Lista dentro do dicionário
    "score": 9.5
}

print(f"O usuário {perfil['nome']} tem as skills: {', '.join(perfil['skills'])}")
print(f"Tipo do objeto principal: {type(perfil)}")
```

|Tipo|Nome|Exemplo|Uso do input()|
|----|----|-----|------|
|String|str|"Olá"|input() puro|
|Inteiro|int|10|int(input())|
|Float|float|10.5|float(input())|
|Booleano|bool|True|Lógica de comparação (Ex: x > 0)|
|Lista|list|[1, 2]|input().split()|
|Dicionário|dict|"{'id': 1}"|Estruturas complexas de dados|

### Desafio 01

```python
nome = input("Digite o seu nome: ").strip().title()

if not nome:
    nome = "Visitante"

print("\n" + "="*30)
print(f"Bem-vindo(a), {nome}!")
print("="*30)
```

### Desafio 02

```python
print("--- Data de Nascimento ---")

try:
    dia = input("Digite o dia (DD): ").strip().zfill(2)
    mes = input("Digite o mês (MM): ").strip().zfill(2)
    ano = input("Digite o ano (AAAA): ").strip()

    d, m, a = int(dia), int(mes), int(ano)

    if not (1 <= d <= 31 and 1 <= m <= 12 and 1900 <= a <= 2026):
        print("Atenção: Esta data parece inválida, mas vou formatá-la.")
    print("\nData formatada com sucesso:")
    print(f"{dia}/{mes}/{ano}")
except ValueError:
    print(" Erro: Por favor, use apenas números para dia, mês e ano.")
```

### Desafio 03

```python
print("="*30)
print("            SOMA     ")
print("="*30)

try:
    num1 = float(input(" -> Digite o 1º número: ").replace(',', '.'))
    num2 = float(input(" -> Digite o 2º número: ").replace(',', '.'))

    soma = num1 + num2

    print("-" * 30)
    print(f"A soma de {num1:g} + {num2:g} é: {soma:g}")
    print("="*30)

except ValueError:
    print("\nErro: Por favor, digite apenas números!")
```
