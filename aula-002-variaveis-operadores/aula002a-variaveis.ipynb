### Desafio 02

```python
print("--- Data de Nascimento ---")

try:
    dia = input("Digite o dia (DD): ").strip().zfill(2)
    mes = input("Digite o mês (MM): ").strip().zfill(2)
    ano = input("Digite o ano (AAAA): ").strip()

    d, m, a = int(dia), int(mes), int(ano)

    if not (1 <= d <= 31 and 1 <= m <= 12 and 1900 <= a <= 2026):
        print("Atenção: Esta data parece inválida, mas vou formatá-la.")
    print("\nData formatada com sucesso:")
    print(f"{dia}/{mes}/{ano}")
except ValueError:
    print(" Erro: Por favor, use apenas números para dia, mês e ano.")
```

Converta uma temperatura de Fahrenheit para Celsius. A fórmula de conversão é: 
`Celsius = (Fahrenheit- 32) * 5/9`

```python
fahrenheit = float(input("Digite a temperatura em Fahrenheit: "))

celsius = (fahrenheit - 32) * 5/9

print(f"{fahrenheit}°F equivalem a {celsius:.2f}°C")
```

Você comprou 3 livros por R$ 25 cada e 2 canetas por R$ 5 cada. Calcule o total gasto.

```python
preco_livro = 25
preco_caneta = 5

quantidade_livros = 3
quantidade_canetas = 2

total = (preco_livro * quantidade_livros) + (preco_caneta * quantidade_canetas)

print(f"Total gasto: R$ {total}")
```

Um carro percorreu 150 km a uma velocidade média de 60 km/h. Quanto tempo (em horas) o carro levou para percorrer essa distância?

```python
distancia = 150  # km
velocidade = 60  # km/h

# Cálculo do tempo
tempo = distancia / velocidade

print(f"O carro levou {tempo} horas para percorrer {distancia} km.")
```

Leia 2 valores A e B, que correspondem a 2 notas de um aluno. A seguir calcule e informe a média aritmética do aluno.

```python
nota_a = float(input("Digite a primeira nota (A): "))
nota_b = float(input("Digite a segunda nota (B): "))

media = (nota_a + nota_b) / 2

print(f"A média do aluno é: {media:.2f}")
```

Leia 2 valores A e B, que correspondem a 2 notas de um aluno. A seguir calcule e informe a média ponderada do aluno, sabendo que a nota A tem peso 4 e a nota B tem peso 6.

Exemplo: nota a * 4 e nota b * 6.

```python
nota_a = float(input("Digite a nota A: "))
nota_b = float(input("Digite a nota B: "))

peso_a = 4
peso_b = 6

media_ponderada = (nota_a * peso_a + nota_b * peso_b) / (peso_a + peso_b)

print(f"A média ponderada do aluno é: {media_ponderada:.2f}")
```

Neste problema, deve-se ler o nome de uma peça que chamaremos de peça1, o número de peças1 que o usuário quer, o valor unitário de cada peça1, o nome de uma peça2, o número de peças2 e o valor unitário de cada peça2. Após, calcule e mostre o valor a ser pago.

```python
peca1 = input("Digite o nome da primeira peça: ")
quantidade1 = int(input(f"Digite a quantidade de {peca1}: "))
valor_unitario1 = float(input(f"Digite o valor unitário de {peca1}: R$ "))

peca2 = input("Digite o nome da segunda peça: ")
quantidade2 = int(input(f"Digite a quantidade de {peca2}: "))
valor_unitario2 = float(input(f"Digite o valor unitário de {peca2}: R$ "))

total = (quantidade1 * valor_unitario1) + (quantidade2 * valor_unitario2)

print(f"\nResumo da compra:")
print(f"{peca1}: {quantidade1} x R$ {valor_unitario1:.2f}")
print(f"{peca2}: {quantidade2} x R$ {valor_unitario2:.2f}")
print(f"\nValor total a pagar: R$ {total:.2f}")
```
